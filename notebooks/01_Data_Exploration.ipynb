```python
clean_df = df.copy()
```

```python
columns_to_remove = [
    "Product Description",
    "Product Image",
    "Customer Password",
    "Customer Street",
    "Customer Email"
]

clean_df = clean_df.drop(columns=columns_to_remove)
```

```python
clean_df.shape
```

```python
clean_df[['order date (DateOrders)', 'shipping date (DateOrders)']].head()
```

```python
clean_df[['order date (DateOrders)', 'shipping date (DateOrders)']].dtypes
```

```python
clean_df['order date (DateOrders)'] = pd.to_datetime(
    clean_df['order date (DateOrders)']
)

clean_df['shipping date (DateOrders)'] = pd.to_datetime(
    clean_df['shipping date (DateOrders)']
)
```

```python
clean_df[['order date (DateOrders)', 'shipping date (DateOrders)']].dtypes
```

```python
clean_df['Order_Year'] = clean_df['order date (DateOrders)'].dt.year
```

```python
clean_df['Order_Month'] = clean_df['order date (DateOrders)'].dt.month
```

```python
clean_df['Order_Month_Name'] = clean_df['order date (DateOrders)'].dt.month_name()
```

```python
clean_df['Order_Day'] = clean_df['order date (DateOrders)'].dt.day_name()
```

```python
clean_df[['Order_Year','Order_Month','Order_Month_Name','Order_Day']].head()
```

```python
clean_df.isnull().sum()[clean_df.isnull().sum() > 0]
```

```python
clean_df['Customer Lname'] = clean_df['Customer Lname'].fillna('Unknown')
```

```python
clean_df['Customer Zipcode'] = clean_df['Customer Zipcode'].fillna(0)
```

```python
clean_df = clean_df.drop(columns=['Order Zipcode'])
```
